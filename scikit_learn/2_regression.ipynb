## Ordinary Least Squares (LinearRegression)  
- **Theory:** minimize mean squared error  
  $$
    \hat\beta = \arg\min_\beta \|y - X\beta\|_2^2
    \quad\Longrightarrow\quad
    \hat\beta = (X^\top X)^{-1} X^\top y
  $$
- **Code:** fit `LinearRegression`, inspect coefficients, plot residuals.

## Polynomial Regression  
- **Theory:** map $x \mapsto (1, x, x^2, \dots, x^d)$ and apply OLS  
- **Code:** `PolynomialFeatures` + `LinearRegression` on a nonlinear function.

## Regularized Linear Models (Ridge, Lasso, ElasticNet)  
- **Theory:**  
  - Ridge: $$\min_\beta \|y - X\beta\|_2^2 + \alpha\|\beta\|_2^2$$  
  - Lasso: $$\min_\beta \|y - X\beta\|_2^2 + \alpha\|\beta\|_1$$  
  - ElasticNet: combination of both penalties.  
- **Code:** coefficient paths vs.\ $\alpha$, `GridSearchCV` to select $\alpha$.

## Robust Regression (RANSAC, HuberRegressor, TheilSen)  
- **Theory:** resistant to outliers via inlier selection (RANSAC) or robust loss (Huber).  
- **Code:** compare OLS vs.\ robust on data with synthetic outliers.

## Support Vector Regression (SVR)  
- **Theory:** $\varepsilon$‑insensitive loss, maximize margin in feature space, kernel trick.  
- **Code:** `SVR` with linear & RBF kernels, vary $C$ and $\varepsilon$.

## K‑Nearest Neighbors Regression  
- **Theory:** predict by averaging the targets of the $k$ nearest points.  
- **Code:** `KNeighborsRegressor`, illustrate effect of different $k$ on bias–variance.

## Tree‑Based Regression  
- **DecisionTreeRegressor:** piecewise constant prediction.  
- **RandomForestRegressor & GradientBoostingRegressor:** ensembles, out‑of‑bag error, feature importances.  
- **HistGradientBoostingRegressor:** faster, better for large datasets.

## Gaussian Process Regression  
- **Theory:** non‑parametric Bayesian regression, posterior mean & covariance via kernels.  
- **Code:** `GaussianProcessRegressor`, plot predictive mean and 95% confidence bands.

## Model Evaluation & Hyperparameter Tuning  
- **Metrics:** MSE, RMSE, MAE, $R^2$.  
- **Tools:** cross‑validation, learning curves, `GridSearchCV`, `RandomizedSearchCV`, `HalvingGridSearchCV`.

## Introduction & Setup

In this section, we’ll create a synthetic regression dataset (noisy sine wave), visualize it, and perform a train/test split to prepare for model fitting.


```python
# Imports and Data Generation
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

# 1) Generate synthetic data
np.random.seed(69)
X = np.linspace(-5, 5, 200)
y = np.sin(X) + 0.3 * np.random.randn(len(X))

# 2) Put into a DataFrame for easy inspection
df = pd.DataFrame({'X': X, 'y': y})
display(df.head())

# 3) Visualize the dataset
plt.figure(figsize=(6, 4))
plt.scatter(df['X'], df['y'], s=20, edgecolors='k')
plt.title("Synthetic Noisy Sine Wave")
plt.xlabel("X")
plt.ylabel("y")
plt.grid(True)
plt.show()

# 4) Train/Test split (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(
    df[['X']], df['y'], test_size=0.2, random_state=42
)

print(f"Train size: {X_train.shape[0]} samples")
print(f"Test size:  {X_test.shape[0]} samples")
```
